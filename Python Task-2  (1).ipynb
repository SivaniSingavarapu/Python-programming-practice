## While Loop

### 1.1 Example-1 sum of numbers form 1 to 100

```python
sum  = 1
i = 0
while sum <=100:
    i= i + sum
    sum = sum + 1
print("Sum of numbers from 1 to 100 is:", i)
```

### 1.2 Example-2 factorial of the number

```python
user_input = int(input("Enter the number: "))
factorial = 1
n=1
while n <= user_input:
    factorial = factorial * n
    n= n + 1
print("Factorial is:", factorial)
    
```

## User Defined Functions with For Loop and IF Condition

### Example 1

```python
def Results(Student):
    for Marks in Student:
        if Marks >= 93:
           print("First Class")
        else:
            print("Second Class")
Student = [80,93,65,90,98]
Results(Student)             
```

### Example 2

```python
def Dmart(amount):
    for bill in amount:
        if bill>=1000:
            print(bill,"-Discount Applied")
        else:
            print(bill,"-No Discount")
Shopping_list = [500,1200,800,1500]
Dmart(Shopping_list)
```

## break

 ### Example 1

```python
School_hours = [9,10,11,12,1,2,3]
for Student in School_hours:
    if Student == 12:
       print("Lunch Break")
       break
    print("Class is going on at" ,Student,"o'clock")
    
    
```

### Example 2

```python
shop = ["Rice","Milk","Soap","Oil","Ball"]
for item in shop:
    if item == "Oil":
        print("Oil Found")
        break
    print(item)
```

## continue

### Example 1

```python
school_time = int(input("Enter the school time:"))
for Student in range(school_time,school_time + 1):
    if Student == 12:
        print("Lunch Break")
        continue
    print("Class is going on")
```

### Example 2

```python
for number in range(1, 9):
    if number == 6:
        continue
    print(number)
```
