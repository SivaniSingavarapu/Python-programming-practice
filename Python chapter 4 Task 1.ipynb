### 2.Create a list of numbers and print only even numbers using a for loop.

```python
even_numbers =[]
numbers = range(1,80)
for i in numbers:
    if i%2==0:
        even_numbers.append(i)
print(even_numbers)
```

### 3.Write a program that handles division by zero using try-except.

```python
num = int(input("Enter Numerator:"))
den = int(input("Enter Denominator:"))
try:
    answer = num/den
    print(answer)
except ZeroDivisionError:
    print("Error:Cannot divided by Zero.")
```

### 4.Given a string, count how many vowels it has.

```python
user_input = input("Enter any Word or Sentence:")
count = 0
for i in user_input.lower():
    if i in "aeiou":
        count+=1
print("Total Number of vowels are:",count)
```

### 5.Reverse a list without using built-in reverse functions.

```python
original_list = [1,2,3,4,5]
updated_list =[]
i = len(original_list)-1
while i>=0:
    updated_list.append(original_list[i])
    i=i-1
updated_list
```
