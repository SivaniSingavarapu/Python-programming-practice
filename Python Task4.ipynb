## Word guessing game

```python
def word_guessing_game():
    import random
    print("Welcome to Word Guessing Game!!!")
    words = ["python","machine","learning","bridge"]
    secret_word = random.choice(words)
    guessed_letters = []
    wrong_attempts = 6
    while wrong_attempts > 0:
        display_word = ""
        for letter in secret_word:
            if letter in guessed_letters:
                display_word += letter + " "
            else:
                display_word += "_"
        print("\nword:", display_word)
        if "_" not in display_word:
            print("Congratulations! You guessed the word:", secret_word)
            break
        try:
            user_input = input("Enter one letter :").lower()
            if len(user_input) != 1 or not user_input.isalpha():
               print("please enter only one alphabet letter.")
               continue
            guessed_letters.append(user_input)
            if user_input in secret_word:
                print("Correct Guess!")
            else:
                wrong_attempts -= 1
                print("Wrong Guess!")
                print("Remaining Attempts:",wrong_attempts)
        except ValueError:
            print("Invalid Input!")
    if"_" not in display_word:
       print("congratulations! You Won the Game.")
    else:
        print("Sorry! All your chances are over.")
        print("The correct word is:",secret_word)
word_guessing_game()
    
```
